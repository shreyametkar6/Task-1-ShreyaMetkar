```python
import json
import os

FILE_NAME = "tasks.json"

# ---------- MODEL (data logic) ----------
def load_tasks():
    if os.path.exists(FILE_NAME):
        with open(FILE_NAME, "r") as file:
            return json.load(file)
    return []

def save_tasks(tasks):
    with open(FILE_NAME, "w") as file:
        json.dump(tasks, file, indent = 4)

def add_task(tasks, text):
    row = {"id" : len(tasks) + 1, "task" : text}  
    tasks.append(row)                           
    save_tasks(tasks)

# ---------- VIEW (user interface) ----------
def view_tasks(tasks):
    if not tasks:
        print("\nNo tasks yet.")
        return
    for row in tasks:
        print(f"{row['id']}. {row['task']}")

def show_menu():
    print("\n--- TO-DO LIST ---")
    print("1. Add Task")
    print("2. View Tasks")
    print("3. Exit")

# ---------- MAIN ----------
def main():
    tasks = load_tasks()

    while True:
        show_menu()
        choice = input("\nChoose an option: ").strip()

        if choice == "1":
            text = input("Enter a task: ").strip()
            if text:
                add_task(tasks, text)
                print("Task added!")
            else:
                print("Task cannot be empty.")
                
        elif choice == "2":
            view_tasks(tasks)
            
        elif choice == "3":
            print("Goodbye!")
            break
            
        else:
            print("Invalid choice, try again.")

if __name__ == "__main__":
    main()
```
